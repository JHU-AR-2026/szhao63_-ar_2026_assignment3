```python
import numpy as np


def camera_to_pixel(
    point,
    W,
    H,
    fx,
    fy,
    ox,
    oy,
    near,
    far
):
    # OpenGL projection matrix
    M_proj = np.array([
        [
            2.0 * fx / W,
            0.0,
            (W - 2.0 * ox) / W,
            0.0
        ],
        [
            0.0,
            2.0 * fy / H,
            (-H + 2.0 * oy) / H,
            0.0
        ],
        [
            0.0,
            0.0,
            -(far + near) / (far - near),
            -(2.0 * far * near) / (far - near)
        ],
        [
            0.0,
            0.0,
            -1.0,
            0.0
        ]
    ])

    # Camera coordinates -> clip coordinates
    p_camera = np.array([
        point[0],
        point[1],
        point[2],
        1.0
    ])

    p_clip = M_proj @ p_camera

    # Perspective divide: clip -> NDC
    p_ndc = p_clip[:3] / p_clip[3]

    # NDC -> pixel coordinates
    u = 0.5 * (p_ndc[0] + 1.0) * W
    v = 0.5 * (p_ndc[1] + 1.0) * H

    return M_proj, p_clip, p_ndc, np.array([u, v])


# Given values
W = 800
H = 600

fx = 800
fy = 800
ox = 400
oy = 400

near = 0.2
far = 20.0

point = (0.3, -0.15, -6.0)

M_proj, p_clip, p_ndc, pixel = camera_to_pixel(
    point,
    W,
    H,
    fx,
    fy,
    ox,
    oy,
    near,
    far
)

print("Projection matrix:")
print(M_proj)

print("\nClip coordinates:")
print(p_clip)

print("\nNDC coordinates:")
print(p_ndc)

print("\nPixel coordinates:")
print(pixel)
```
